```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
X=[
    [0,0],
    [0,1],
    [1,0],
    [1,1]
]
y=[0,0,0,1]
```

```python
def train_preceptron(inputs,targets):
    w1,w2,b=0.0,0.0,0.0
    lr=0.1

    for epoch in range(20):
        errors=0
        for i in range(len(inputs)):
            x1,x2=inputs[i][0],inputs[i][1]
            target=targets[i]

            # Calculate weighted sum
            z=(w1*x1)+(w2*x2)+b
            #Apply activation function
            pred=1 if z>=0 else 0

            # Update weights if prediction is incorrect
            if target!=pred:
                error=target-pred
                w1+=lr*error*x1
                w2+=lr*error*x2
                b+=lr*error

                errors+=1

        #stop if no errors
        if errors==0:
            break
    return w1,w2,b
```

```python
w1,w2,b=train_preceptron(X,y)
print(f"Weights: [{w1:.1f}, {w2:.1f}], Bias: {b:.1f}")
```

# The decision boundary is the line where:

$$w_1x_1 + w_2x_2 + b = 0$$

```python
plt.figure(figsize=(6,5))
for i in range(len(X)):
    color='red' if y[i]==1 else 'blue'
    plt.scatter(
        X[i][0],
        X[i][1],
        color=color,
        s=200,
        edgecolors='black',
        zorder=3
    )

#calculate decision boundry

x1_line=[-0.5,1.5]
x2_line=[
    -(w1*x+b)/w2
    for x in x1_line
]
plt.plot(
    x1_line,
    x2_line,
    color='black',
    linewidth=2.5,
    label="Decision Boundry",
    
)

# Step 6: Graph formatting
plt.xlim(-0.5,1.5)
plt.ylim(-0.5,1.5)
plt.xlabel("X1")
plt.ylabel("X2")
plt.title("Perceptron Boundry Line")
plt.grid(True,alpha=0.3)
plt.legend()
plt.show()

    
```
